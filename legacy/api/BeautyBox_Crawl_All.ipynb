```python
from pathlib import Path
REPO_ROOT = next(
    p for p in (Path.cwd(), *Path.cwd().parents)
    if (p / "beautybox_data.py").is_file() and (p / "legacy/api").is_dir()
)
LEGACY_SCRIPT = REPO_ROOT / "legacy/api/crawl_all_beautybox.py"
print("Script API cũ:", LEGACY_SCRIPT)
```

```python
import subprocess, sys
OUTPUT = REPO_ROOT / "data/runs/legacy_api_current"
RESUME = False
command = [sys.executable, str(LEGACY_SCRIPT), "--output", str(OUTPUT)]
if RESUME:
    command.append("--resume")
subprocess.run(command, check=False)
```

```python
import json
status = json.loads((Path(OUTPUT) / "status.json").read_text(encoding="utf-8"))
print(json.dumps(status, ensure_ascii=False, indent=2))
if status["status"] != "complete":
    print("CHƯA HOÀN TẤT: đọc errors và missing_product_ids trước khi sử dụng.")
```

```python
import shutil
archive = shutil.make_archive(str(OUTPUT), "zip", str(OUTPUT))
print("File kết quả:", archive)
```
